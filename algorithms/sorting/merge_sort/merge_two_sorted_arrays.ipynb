Write a function:

```python
def merge_two_sorted_arrays(left, right):
```

Functionalities:

the function should merge two already sorted arrays into one sorted array.

Return rules:

- Return a new sorted array.
- Do not use `sort()` or `sorted()`

Example:

```
Input: left = [1, 3, 5], right = [2, 4, 6]
Output: [1, 2, 3, 4, 5, 6]

Input:  left = [1, 2, 7], right = [3, 4]
Output: [1, 2, 3, 4, 7]
```

````python
def merge_two_sorted_arrays(left, right):
    ans = []

    i, j = 0, 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            ans.append(left[i])
            i += 1
        else:
            ans.append(right[j])
            j += 1
    
    while i < len(left):
        ans.append(left[i])
        i += 1
    
    while j < len(right):
        ans.append(right[j])
        j += 1

    return ans
````

````python
# Test 1
left = [1, 3, 5]
right = [2, 4, 6]

print(merge_two_sorted_arrays(left, right))
````

````python
# Test 2
left = [1, 2, 7]
right = [3, 4]

print(merge_two_sorted_arrays(left, right))
````
