## Task: Implement Merge Sort (Index Version)

Write a function:

```python
def merge_sort(A, left, right):
```

You must:

- Use index (left, right) instead of slicing
- Implement merge(A, left, mid, right)
- Modify the original array

Functionalities:
- Sort the array in-place
- Use recursion
- Follow pseudocode structure

Example:

```
Input:  A = [5, 2, 8, 1]
After:  A = [1, 2, 5, 8]
```

````python
def merge_sort(A, left, right):
    if left >= right:
        return

    mid = (left + right) // 2

    merge_sort(A, left, mid)
    merge_sort(A, mid+1, right)

    merge(A, left, mid, right)


def merge(A, left, mid, right):
    temp = []

    i = left
    j = mid + 1

    while i <= mid and j <= right:
        if A[i] <= A[j]:
            temp.append(A[i])
            i += 1
        else:
            temp.append(A[j])
            j += 1
    
    while i <= mid:
        temp.append(A[i])
        i += 1
    
    while j <= right:
        temp.append(A[j])
        j += 1

    for k in range(len(temp)):
        A[left + k] = temp[k]
````

````python
# Test
A = [5, 2, 8, 1]
merge_sort(A, 0, len(A)-1)

print(A)
````
