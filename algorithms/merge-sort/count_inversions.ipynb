## Task: Count Inversions Using Merge Sort

### Write a function:

```python
def count_inversions(arr):
```

---

### Functionalities:

An inversion is a pair (i, j) where:

```python
i < j and arr[i] > arr[j]
```

The function should count how many inversions are in the array.

---

### Return rules:

- Return the number of inversions.
- Do not use brute force nested loops.
- Use merge sort idea.

---

### Example:

```
Input:  [2, 4, 1, 3, 5]
Output: 3
```

Because the inversions are:

```
(2, 1), (4, 1), (4, 3)
```

````python
def count_inversions(arr):
    sorted_arr, count = merge_sort_and_count(arr)
    return count


def merge_sort_and_count(arr):
    if len(arr) <= 1:
        return arr, 0
    
    mid = len(arr) // 2

    left = arr[:mid]
    right = arr[mid:]

    sorted_left, count_left = merge_sort_and_count(left)
    sorted_right, count_right = merge_sort_and_count(right)

    merged_arr, count_cross = merge_and_count(left, right)

    total_count = count_left + count_right + count_cross

    return merged_arr, total_count


def merge_and_count(left, right):
    result = []
    count = 0

    i, j = 0, 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            count += len(left) - i
            j += 1
        
    while i < len(left):
        result.append(left[i])
        i += 1
    
    while j < len(right):
        result.append(right[j])
        j += 1
    
    return result, count
````

````python
# Test
arr = [2, 4, 1, 3, 5]
print(count_inversions(arr))
````
